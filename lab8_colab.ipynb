```python
%%writefile shop.py
"""Модуль для расчётов в интернет-магазине."""


def total(prices):
    """Возвращает сумму цен, округлённую до 2 знаков."""
    return round(sum(prices), 2)


def apply_discount(price, percent):
    """Возвращает цену со скидкой percent процентов."""
    return round(price * (1 - percent / 100), 2)


def most_expensive(items):
    """Возвращает название и цену самого дорогого товара из словаря."""
    name = max(items, key=items.get)
    return name, items[name]


if __name__ == "__main__":
    print(total([100, 250.5, 49.5]))
    print(apply_discount(1000, 15))
    print(most_expensive({"хлеб": 180, "сыр": 2400, "чай": 950}))
```

```python
import shop

cart = {"молоко": 520, "хлеб": 180, "сыр": 2400, "чай": 950}
print(shop.total(cart.values()))
print(shop.apply_discount(shop.total(cart.values()), 10))
print(shop.most_expensive(cart))
```

**Почему в Colab мы не создавали venv?**

Виртуальное окружение нужно, чтобы изолировать пакеты одного проекта от других проектов на *одном и том же* компьютере. В Colab каждый блокнот работает на отдельной временной виртуальной машине: она создаётся для сеанса и удаляется после его окончания вместе со всеми установленными пакетами и файлами (в том числе shop.py). Поэтому смешивать пакеты разных проектов там нечему — сама виртуальная машина уже играет роль изолированного окружения. Именно поэтому ячейки с `%%writefile` и `%pip install` приходится выполнять заново при каждом открытии блокнота.
